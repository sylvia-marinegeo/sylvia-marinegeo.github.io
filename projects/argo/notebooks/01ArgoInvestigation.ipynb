# 03 T-S Diagram

```python
# Select cycle 1, ascending profile
profile = ds.where(
    (ds.CYCLE_NUMBER == 1) &
    (ds.DIRECTION == "A"),
    drop=True
)

# Create a dataframe
ts_data = profile[
    ["PRES", "TEMP", "PSAL", "TEMP_QC", "PSAL_QC"]
].to_dataframe().reset_index()

# Retain good and probably good measurements
ts_data = ts_data[
    ts_data["TEMP_QC"].isin([1, 2]) &
    ts_data["PSAL_QC"].isin([1, 2])
].dropna(subset=["TEMP", "PSAL", "PRES"])

print(ts_data.head())
print("\nNumber of valid measurements:", len(ts_data))
```

```python
# Inspect measurements at selected pressure levels

target_pressures = [50, 100, 200, 500, 750, 1000, 1250, 1500, 1750, 1950]

for target in target_pressures:

    closest = (ts_data["PRES"] - target).abs().idxmin()

    row = ts_data.loc[closest]

    print(
        f"Pressure: {row['PRES']:6.1f} dbar | "
        f"Temperature: {row['TEMP']:5.2f} °C | "
        f"Salinity: {row['PSAL']:6.3f}"
    )
```

```python
import gsw

ts_data["SA"] = gsw.SA_from_SP(
    ts_data["PSAL"],
    ts_data["PRES"],
    ts_data["LONGITUDE"],
    ts_data["LATITUDE"]
)

ts_data["CT"] = gsw.CT_from_t(
    ts_data["SA"],
    ts_data["TEMP"],
    ts_data["PRES"]
)

ts_data["sigma0"] = gsw.sigma0(
    ts_data["SA"],
    ts_data["CT"]
)
```

```python
plt.figure(figsize=(7, 8))

plt.plot(
    ts_data["sigma0"],
    ts_data["PRES"],
    linewidth=2
)

plt.gca().invert_yaxis()

plt.xlabel("Potential Density Anomaly (kg/m³)")
plt.ylabel("Pressure (dbar)")
plt.title("Potential Density Profile – Argo Float 1900857")
plt.grid(alpha=0.3)

plt.show()
```

```python
# Define temperature and salinity ranges
temperature_range = np.linspace(0, 18, 100)
salinity_range = np.linspace(34, 35.2, 100)

# Create a 2D grid
S, T = np.meshgrid(salinity_range, temperature_range)

# Convert practical salinity to absolute salinity
SA_grid = gsw.SA_from_SP(S, 0, 9.738, -40.063)

# Convert in-situ temperature to conservative temperature
CT_grid = gsw.CT_from_t(SA_grid, T, 0)

# Calculate potential density anomaly
sigma_grid = gsw.sigma0(SA_grid, CT_grid)
```

# 04 — Hydrographic Variability Along the Float Trajectory

## 4.1 Extracting Individual Profiles

```python
# Select ascending profiles only
ascending = ds.where(
    ds.DIRECTION == "A",
    drop=True
)

# Convert selected variables to a pandas DataFrame
profiles_df = ascending[
    [
        "CYCLE_NUMBER",
        "PRES",
        "TEMP",
        "PSAL",
        "TEMP_QC",
        "PSAL_QC"
    ]
].to_dataframe().reset_index()

# Keep measurements with acceptable quality-control flags
profiles_df = profiles_df[
    profiles_df["TEMP_QC"].isin([1, 2]) &
    profiles_df["PSAL_QC"].isin([1, 2])
]

# Remove measurements with missing values
profiles_df = profiles_df.dropna(
    subset=["PRES", "TEMP", "PSAL"]
)

# Sort measurements by cycle and pressure
profiles_df = profiles_df.sort_values(
    ["CYCLE_NUMBER", "PRES"]
).reset_index(drop=True)

print("Valid measurements:", len(profiles_df))
print("Number of cycles:", profiles_df["CYCLE_NUMBER"].nunique())

profiles_df.head()
```

```python
# Select intermediate-depth measurements
intermediate = profiles_df[
    profiles_df["PRES"].between(400, 1500)
].copy()

# Find the row containing minimum salinity in each cycle
min_indices = intermediate.groupby("CYCLE_NUMBER")["PSAL"].idxmin()

# Create a table containing one minimum per cycle
salinity_minima = intermediate.loc[min_indices].copy()

# Select the variables we need
salinity_minima = salinity_minima[
    [
        "CYCLE_NUMBER",
        "TIME",
        "LATITUDE",
        "LONGITUDE",
        "PRES",
        "TEMP",
        "PSAL"
    ]
]

# Rename columns for clarity
salinity_minima = salinity_minima.rename(
    columns={
        "PRES": "Min_Salinity_Pressure",
        "TEMP": "Min_Salinity_Temperature",
        "PSAL": "Min_Salinity"
    }
)

# Sort by cycle number
salinity_minima = salinity_minima.sort_values(
    "CYCLE_NUMBER"
).reset_index(drop=True)

print("Profiles analyzed:", len(salinity_minima))

salinity_minima.head(10)
```

```python
# Summary statistics for all salinity minima
print(
    salinity_minima[
        [
            "Min_Salinity",
            "Min_Salinity_Pressure",
            "Min_Salinity_Temperature"
        ]
    ].describe().round(3)
)

# Check whether minima occur near the search boundaries
boundary_profiles = salinity_minima[
    (salinity_minima["Min_Salinity_Pressure"] <= 425) |
    (salinity_minima["Min_Salinity_Pressure"] >= 1475)
]

print("\nProfiles near search boundaries:", len(boundary_profiles))

boundary_profiles[
    ["CYCLE_NUMBER", "Min_Salinity_Pressure", "Min_Salinity"]
]
```

```python
# Extract cycle 160
cycle_160 = profiles_df[
    profiles_df["CYCLE_NUMBER"] == 160
].copy()

# Plot its salinity profile
plt.figure(figsize=(7, 8))

plt.plot(
    cycle_160["PSAL"],
    cycle_160["PRES"],
    linewidth=2
)

plt.axhline(
    400,
    color="red",
    linestyle="--",
    label="Search window begins"
)

plt.axhline(
    1500,
    color="gray",
    linestyle="--",
    label="Search window ends"
)

plt.gca().invert_yaxis()

plt.xlabel("Practical Salinity")
plt.ylabel("Pressure (dbar)")
plt.title("Salinity Profile – Cycle 160")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

### Cycle 60 Investigation

##### A pronounced minimum around 400–450 dbar. Salinity decreases toward approximately 34.22, then begins increasing. There are small fluctuations around the minimum, but it is not simply a continuously decreasing profile.

##### A relatively broad low-salinity region from approximately 400–800 dbar. Salinity fluctuates around 34.25–34.30 before increasing more noticeably below 800 dbar.

##### A strong near-surface salinity minimum. Above approximately 100 dbar, salinity becomes substantially lower than in the intermediate water. This is exactly why we restricted our search to intermediate depths.

## 4.2 — Temperature Variability at 500 dbar

```python
# Create a list to store the results
temp_500 = []

# Process each ascending profile
for cycle, profile in profiles_df.groupby("CYCLE_NUMBER"):

    # Sort measurements by pressure
    profile = profile.sort_values("PRES")

    # Only interpolate if 500 dbar is within the measured range
    if profile["PRES"].min() <= 500 <= profile["PRES"].max():

        temperature = np.interp(
            500,
            profile["PRES"],
            profile["TEMP"]
        )

        temp_500.append({
            "CYCLE_NUMBER": cycle,
            "TIME": profile["TIME"].iloc[0],
            "LATITUDE": profile["LATITUDE"].iloc[0],
            "LONGITUDE": profile["LONGITUDE"].iloc[0],
            "TEMP_500": temperature
        })

# Convert results to a DataFrame
temp_500_df = pd.DataFrame(temp_500)

print("Profiles with valid 500 dbar estimates:", len(temp_500_df))

temp_500_df.head()
```

```python
plt.figure(figsize=(11, 5))

sc = plt.scatter(
    temp_500_df["TIME"],
    temp_500_df["TEMP_500"],
    c=temp_500_df["LATITUDE"],
    cmap="viridis",
    s=45,
    edgecolors="black",
    linewidths=0.3
)

# Highlight cycle 44
cycle_44 = temp_500_df[
    temp_500_df["CYCLE_NUMBER"] == 44
]

plt.scatter(
    cycle_44["TIME"],
    cycle_44["TEMP_500"],
    color="red",
    s=150,
    marker="*",
    label="Cycle 44",
    zorder=5
)

plt.colorbar(sc, label="Latitude (°)")
plt.xlabel("Date")
plt.ylabel("Temperature at 500 dbar (°C)")
plt.title("Temperature Variability at 500 dbar")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
isopycnal_27 = []

for cycle, profile in profiles_df.groupby("CYCLE_NUMBER"):

    profile = profile.sort_values("PRES")

    pressure = profile["PRES"].to_numpy()

    SA = gsw.SA_from_SP(
        profile["PSAL"].to_numpy(),
        pressure,
        profile["LONGITUDE"].to_numpy(),
        profile["LATITUDE"].to_numpy()
    )

    CT = gsw.CT_from_t(
        SA,
        profile["TEMP"].to_numpy(),
        pressure
    )

    sigma0 = gsw.sigma0(SA, CT)

    # Identify where sigma0 crosses 27.0
    crossings = np.where(
        (sigma0[:-1] - 27.0) *
        (sigma0[1:] - 27.0) < 0
    )[0]

    # Keep profiles with exactly one crossing
    if len(crossings) == 1:

        i = crossings[0]

        crossing_pressure = np.interp(
            27.0,
            sigma0[i:i+2],
            pressure[i:i+2]
        )

        isopycnal_27.append({
            "CYCLE_NUMBER": cycle,
            "TIME": profile["TIME"].iloc[0],
            "LATITUDE": profile["LATITUDE"].iloc[0],
            "LONGITUDE": profile["LONGITUDE"].iloc[0],
            "PRES_SIGMA27": crossing_pressure
        })

isopycnal_df = pd.DataFrame(isopycnal_27)

print("Profiles with one σ₀ = 27.0 crossing:", len(isopycnal_df))

isopycnal_df.head()
```

```python
target_date = pd.Timestamp("2009-05-13")
origin = pd.Timestamp("1950-01-01")

target_days = (target_date - origin).days

matching_indices = []

with remote_file as f:
    with h5py.File(f, "r") as h5:

        time_ds = h5["time"]

        scale = float(time_ds.attrs["scale_factor"][0])
        offset = float(time_ds.attrs["add_offset"][0])

        block_size = 1_000_000
        total = time_ds.shape[0]

        for start in range(0, total, block_size):

            end = min(start + block_size, total)

            raw_times = time_ds[start:end]

            decoded_days = raw_times * scale + offset

            matches = np.where(
                np.isclose(
                    decoded_days,
                    target_days,
                    atol=0.001,
                    rtol=0
                )
            )[0]

            matching_indices.extend(start + matches)

            print(
                f"Processed {end:,} / {total:,}",
                end="\r"
            )

matching_indices = np.array(matching_indices, dtype=np.int64)

print("\nMay 13 observations:", len(matching_indices))
```

```python
float_lat = -40.177
float_lon = 20.901

with remote_file as f:
    with h5py.File(f, "r") as h5:

        # Read the coordinates and track IDs for the matching dates
        latitudes = h5["latitude"][matching_indices]
        longitudes = h5["longitude"][matching_indices]
        tracks = h5["track"][matching_indices]

        # Inspect the units before interpreting radius or amplitude
        print("Effective radius attributes:")
        for key, value in h5["effective_radius"].attrs.items():
            if key not in ["DIMENSION_LIST", "_Netcdf4Coordinates"]:
                print(f"  {key}: {value}")

# Calculate great-circle distances from the float
lat1 = np.radians(float_lat)
lat2 = np.radians(latitudes)

dlat = lat2 - lat1
dlon = np.radians(
    (longitudes - float_lon + 180) % 360 - 180
)

a = (
    np.sin(dlat / 2)**2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
)

distances_km = 6371 * 2 * np.arcsin(
    np.sqrt(np.clip(a, 0, 1))
)

nearby_mask = distances_km <= 500

nearby_eddies = pd.DataFrame({
    "Track": tracks[nearby_mask],
    "Latitude": latitudes[nearby_mask],
    "Longitude": longitudes[nearby_mask],
    "Distance_km": distances_km[nearby_mask]
})

nearby_eddies = nearby_eddies.sort_values(
    "Distance_km"
).reset_index(drop=True)

print("\nAnticyclonic eddies within 500 km:")
print(nearby_eddies.to_string(index=False))
```

```python
with remote_file as f:
    with h5py.File(f, "r") as h5:

        radius_ds = h5["effective_radius"]
        amplitude_ds = h5["amplitude"]

        radius_scale = float(
            radius_ds.attrs.get("scale_factor", [1])[0]
        )

        amplitude_scale = float(
            amplitude_ds.attrs.get("scale_factor", [1])[0]
        )

        print("Amplitude attributes:")
        for key, value in amplitude_ds.attrs.items():
            if key not in ["DIMENSION_LIST", "_Netcdf4Coordinates"]:
                print(f"  {key}: {value}")

        radii = (
            radius_ds[matching_indices][nearby_mask]
            * radius_scale / 1000
        )

        amplitudes = (
            amplitude_ds[matching_indices][nearby_mask]
            * amplitude_scale
        )

nearby_eddies = pd.DataFrame({
    "Track": tracks[nearby_mask],
    "Latitude": latitudes[nearby_mask],
    "Longitude": longitudes[nearby_mask],
    "Distance_km": distances_km[nearby_mask],
    "Radius_km": radii,
    "Amplitude": amplitudes
}).sort_values("Distance_km").reset_index(drop=True)

print(nearby_eddies.to_string(index=False))
```

```python
target_densities = [26.0, 26.5, 27.0]

results = []

for cycle in [43, 44, 45]:

    group = (
        profiles_df[
            profiles_df["CYCLE_NUMBER"] == cycle
        ]
        .sort_values("sigma0")
        .dropna(subset=["sigma0", "CT", "SA"])
    )

    # Average any duplicate density values before interpolation
    group = group.groupby("sigma0", as_index=False)[
        ["CT", "SA"]
    ].mean()

    for density in target_densities:

        if group["sigma0"].min() <= density <= group["sigma0"].max():

            ct = np.interp(
                density,
                group["sigma0"],
                group["CT"]
            )

            sa = np.interp(
                density,
                group["sigma0"],
                group["SA"]
            )

            results.append({
                "Cycle": cycle,
                "sigma0": density,
                "CT (°C)": ct,
                "SA (g/kg)": sa
            })

density_comparison = pd.DataFrame(results)

print(
    density_comparison.round(3).to_string(index=False)
)
```

```python
# Extract profiles
profiles = {}

for cycle in [43, 44, 45]:
    profiles[cycle] = (
        profiles_df[
            profiles_df["CYCLE_NUMBER"] == cycle
        ]
        .sort_values("PRES")
    )

# Common pressure grid
pressure_grid = np.arange(25, 1951, 25)

# Interpolate Conservative Temperature
ct_interp = {}

for cycle, group in profiles.items():

    ct_interp[cycle] = np.interp(
        pressure_grid,
        group["PRES"],
        group["CT"]
    )

# Reference: mean of neighboring cycles
ct_reference = (
    ct_interp[43] + ct_interp[45]
) / 2

# Cycle 44 temperature anomaly
ct_anomaly = ct_interp[44] - ct_reference

# Plot
fig, ax = plt.subplots(figsize=(7, 8))

ax.plot(
    ct_anomaly,
    pressure_grid,
    linewidth=2
)

ax.axvline(0, color="black", linestyle="--")

ax.invert_yaxis()

ax.set_xlabel("Conservative Temperature Anomaly (°C)")
ax.set_ylabel("Pressure (dbar)")

ax.set_title(
    "Cycle 44 Temperature Anomaly\n"
    "Relative to Cycles 43 and 45"
)

ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

The temperature anomaly is not confined to the surface or upper thermocline. It extends through much of the upper 1,000 dbar and remains positive down to approximately 2,000 dbar.

```python
# Interpolate Absolute Salinity onto the same pressure grid
sa_interp = {}

for cycle, group in profiles.items():

    sa_interp[cycle] = np.interp(
        pressure_grid,
        group["PRES"],
        group["SA"]
    )

# Reference: average of cycles 43 and 45
sa_reference = (
    sa_interp[43] + sa_interp[45]
) / 2

# Cycle 44 salinity anomaly
sa_anomaly = sa_interp[44] - sa_reference

# Plot
fig, ax = plt.subplots(figsize=(7, 8))

ax.plot(
    sa_anomaly,
    pressure_grid,
    linewidth=2
)

ax.axvline(0, color="black", linestyle="--")

ax.invert_yaxis()

ax.set_xlabel("Absolute Salinity Anomaly (g/kg)")
ax.set_ylabel("Pressure (dbar)")

ax.set_title(
    "Cycle 44 Salinity Anomaly\n"
    "Relative to Cycles 43 and 45"
)

ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Interpolate potential density onto the common pressure grid
sigma_interp = {}

for cycle, group in profiles.items():

    sigma_interp[cycle] = np.interp(
        pressure_grid,
        group["PRES"],
        group["sigma0"]
    )

# Reference: average of cycles 43 and 45
sigma_reference = (
    sigma_interp[43] + sigma_interp[45]
) / 2

# Cycle 44 density anomaly
sigma_anomaly = sigma_interp[44] - sigma_reference

# Plot
fig, ax = plt.subplots(figsize=(7, 8))

ax.plot(
    sigma_anomaly,
    pressure_grid,
    linewidth=2
)

ax.axvline(0, color="black", linestyle="--")

ax.invert_yaxis()

ax.set_xlabel(
    r"Potential Density Anomaly Difference (kg/m³)"
)
ax.set_ylabel("Pressure (dbar)")

ax.set_title(
    "Cycle 44 Potential Density Difference\n"
    "Relative to Cycles 43 and 45"
)

ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Determine the range of atlas indices
first_idx = int(matched_contours["Atlas_Index"].min())
last_idx = int(matched_contours["Atlas_Index"].max())

print(f"Retrieving contours {first_idx}–{last_idx}")

# Read the entire interval in two operations
with remote_file as f:
    with h5py.File(f, "r") as h5:

        lat_ds = h5["effective_contour_latitude"]
        lon_ds = h5["effective_contour_longitude"]

        contour_lat_block = lat_ds[first_idx:last_idx + 1]
        contour_lon_block = lon_ds[first_idx:last_idx + 1]

        lat_scale = float(np.asarray(
            lat_ds.attrs["scale_factor"]
        ).item())

        lat_offset = float(np.asarray(
            lat_ds.attrs["add_offset"]
        ).item())

        lon_scale = float(np.asarray(
            lon_ds.attrs["scale_factor"]
        ).item())

        lon_offset = float(np.asarray(
            lon_ds.attrs["add_offset"]
        ).item())

# Convert packed coordinates to geographic coordinates
contour_lat_block = (
    contour_lat_block.astype(float) * lat_scale + lat_offset
)

contour_lon_block = (
    contour_lon_block.astype(float) * lon_scale + lon_offset
)

print("Latitude array shape:", contour_lat_block.shape)
print("Longitude array shape:", contour_lon_block.shape)
print("Contour data retrieved successfully.")
```

```python
from matplotlib.path import Path

containment_results = []

for _, row in matched_contours.iterrows():

    idx = int(row["Atlas_Index"])

    # Locate the contour within our downloaded arrays
    local_idx = idx - first_idx

    contour_lat = contour_lat_block[local_idx]
    contour_lon = contour_lon_block[local_idx]

    # Remove invalid coordinates, if present
    valid = (
        np.isfinite(contour_lat)
        & np.isfinite(contour_lon)
        & (contour_lat >= -90)
        & (contour_lat <= 90)
        & (contour_lon >= -180)
        & (contour_lon <= 360)
    )

    vertices = np.column_stack([
        contour_lon[valid],
        contour_lat[valid]
    ])

    # Test whether float position is inside the contour
    if len(vertices) >= 3:
        polygon = Path(
            np.vstack([vertices, vertices[0]])
        )

        inside = bool(
            polygon.contains_point(
                (row["LONGITUDE"], row["LATITUDE"])
            )
        )
    else:
        inside = np.nan

    containment_results.append({
        "Cycle": int(row["CYCLE_NUMBER"]),
        "Date": row["TIME"].date(),
        "Inside_Eddy": inside
    })

containment_df = pd.DataFrame(containment_results)

print(containment_df.to_string(index=False))

print(
    "\nProfiles inside eddy:",
    containment_df["Inside_Eddy"].eq(True).sum(),
    "of",
    len(containment_df)
)
```

```python
target_sigma = 27.0
cycles = [43, 44, 45]

results = []

for cycle in cycles:

    profile = profiles_df[
        profiles_df["CYCLE_NUMBER"] == cycle
    ][["PRES", "sigma0"]].copy()

    profile = (
        profile
        .dropna()
        .sort_values("PRES")
    )

    pressure = profile["PRES"].to_numpy()
    density = profile["sigma0"].to_numpy()

    # Find every interval crossing the target density
    crossings = np.where(
        (density[:-1] - target_sigma) *
        (density[1:] - target_sigma) <= 0
    )[0]

    crossing_pressures = []

    for i in crossings:

        if density[i + 1] == density[i]:
            continue

        p = np.interp(
            target_sigma,
            [density[i], density[i + 1]],
            [pressure[i], pressure[i + 1]]
        ) if density[i] <= density[i + 1] else (
            pressure[i] +
            (target_sigma - density[i]) *
            (pressure[i + 1] - pressure[i]) /
            (density[i + 1] - density[i])
        )

        crossing_pressures.append(p)

    results.append({
        "Cycle": cycle,
        "Crossings_dbar": crossing_pressures,
        "Deepest_crossing_dbar": (
            max(crossing_pressures)
            if crossing_pressures else np.nan
        )
    })

isopycnal_df = pd.DataFrame(results)

print(isopycnal_df.to_string(index=False))
```
